1. A bakery tracks the daily sales of muffins (in dozens) over a week: [10, 12, 11, 15, 14, 13, 12].
What is the most representative value of their weekly sales, and why?

Ans:- The most representative value is the mean (average).

Step 1: Add all daily sales.

10 + 12 + 11 + 15 + 14 + 13 + 12 = 87

Step 2: Divide by the number of days (7).
\[
\text{Mean} = \frac{87}{7} = 12.43
\]

Answer: 12.43 dozen muffins
Why?

The mean is the most representative value because it considers the sales of all seven days and gives a good overall picture of the bakery's typical daily sales.

Therefore, the bakery sells approximately 12.43 dozen muffins per day on average.

9. A family records their monthly electricity usage (in kWh): [400, 420, 390, 450, 410]. Find the mean and standard deviation. What do these values together tell you about the family’s energy use
pattern?

Ans:- Let's calculate the **mean and standard deviation** step by step.

### **Step 1: Calculate the mean**

Given electricity usage:

400, 420, 390, 450, 410 kWh

Add all the values:

\\[ 400+420+390+450+410=2070 \\]

Divide by the number of months (5):

\\[ \text{Mean}=\frac{2070}{5}=414 \\]

Mean = 414 kWh

### **Step 2: Calculate the standard deviation**

First, find the difference between each value and the mean, then square each difference.

| Usage (x) | x − 414 | (x − 414)² |
| --------- | ------- | ---------- |
| 400       | -14     | 196        |
| 420       | 6       | 36         |
| 390       | -24     | 576        |
| 450       | 36      | 1,296      |
| 410       | -4      | 16         |
| **Total** | **0**   | **2,120**  |

Using the population standard deviation formula:

\\[ \sigma=\sqrt{\frac{\sum(x-\bar{x})^2}{N}} \\]

Substitute the values:

\\[ \sigma=\sqrt{\frac{2120}{5}} \\]

\\[ =\sqrt{424}=20.59 \\]

Standard deviation = 20.59 kWh

### **Step 3: What do these values tell us?**

- **Mean (414 kWh):** The family uses an average of 414 kWh of electricity per month.
- **Standard deviation (20.59 kWh):** Monthly electricity usage generally varies by about 20.59 kWh around the average.
- **Overall pattern:** The family's electricity consumption is relatively consistent, with only moderate variations from month to month.

### **Final Answer**

Mean

**414 kWh**

Standard deviation

**20.59 kWh**

**Conclusion:** The family consumes approximately 414 kWh of electricity per month, and the relatively low standard deviation indicates a fairly stable energy-use pattern.
