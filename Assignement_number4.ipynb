4.  What is the concept of standardization (z-score), and why is it important in data analysis? Explain the
formula and how standardization transforms a dataset

Ans:-  ## **1. What is standardization (Z-score)?**

**Standardization** is a data transformation technique used to convert values in a dataset into a common scale.

It transforms data so that:

- The mean becomes **0**.
- The standard deviation becomes **1**.

The standardized value is called a **Z-score**.

A Z-score tells us how many standard deviations a particular value is above or below the mean.

## **2. Formula for standardization**

The formula is:

\\[ Z=\frac{X-\mu}{\sigma} \\]

Where:

- \\(Z\\) = Standardized value (Z-score)
- \\(X\\) = Original data value
- \\(\mu\\) = Mean of the dataset
- \\(\sigma\\) = Standard deviation of the dataset

## **3. Example of standardization**

Suppose a dataset has:

- Mean = 50
- Standard deviation = 10
- Original value = 70

Using the formula:

\\[ Z=\frac{70-50}{10} \\]

\\[ Z=\frac{20}{10}=2 \\]

**Interpretation:** The value 70 is 2 standard deviations above the mean.

## **5. Why is standardization important in data analysis?**

Standardization is important for several reasons:

1\. **Comparing different variables:** It allows us to compare variables measured in different units, such as height, weight, and income.

## **2. Formula of Kurtosis**

The formula for population kurtosis is:

\\[ K=\frac{E[(X-\mu)^4]}{\sigma^4} \\]

Where:

- \\(K\\) = Kurtosis
- \\(X\\) = Individual data value
- \\(\mu\\) = Mean
- \\(\sigma\\) = Standard deviation
- \\(E\\) = Expected value

**Excess kurtosis:**

\\[ \text{Excess Kurtosis}=K-3 \\]

For excess kurtosis:

- Positive = Leptokurtic
- Zero = Mesokurtic
- Negative = Platykurtic

6.  Explain why the uniform distribution is a good model for the outcome of rolling a fair die.

Ans:- ## **Why is the Uniform Distribution a Good Model for Rolling a Fair Die?**

A **uniform distribution** is a probability distribution in which all possible outcomes have an equal probability of occurring.

A fair six-sided die has six possible outcomes:

\\[ X=\\{1,2,3,4,5,6\\} \\]

Since the die is fair, each number has an equal chance of appearing.

### **1. Probability of Each Outcome**

The probability of getting any particular number is:

\\[ P(X=x)=\frac{1}{6} \\]

Therefore:

| Outcome | Probability  |
| ------- | ------------ |
| 1       | 1/6 = 16.67% |
| 2       | 1/6 = 16.67% |
| 3       | 1/6 = 16.67% |
| 4       | 1/6 = 16.67% |

1. **Equal probability:** Every number from 1 to 6 has an equal chance of appearing.

4. **Total probability:** The sum of all probabilities is equal to 1.

7.  Use Excel to compute the probability of getting at least 8 successes in 15 trials with success probability 0.5

Ans:- To calculate the probability of getting **at least 8 successes in 15 trials**, we use the **Binomial Distribution** in Excel.

### **Step 1: Given information**

- Number of trials (n) = 15
- Probability of success (p) = 0.5
- Number of successes (X) ≥ 8

### **Step 2: Apply the binomial formula**

We need to calculate:

\\[ P(X\geq 8) \\]

Using the complement rule:

\\[ P(X\geq 8)=1-P(X\leq 7) \\]

### **Step 3: Excel formula**

Enter this formula in Excel:

```
=1-BINOM.DIST(7,15,0.5,TRUE)
```

Explanation:

- `7` = Maximum number of failures in the cumulative calculation.
- `15` = Total number of trials.
- `0.5` = Probability of success.
- `TRUE` = Calculates cumulative probability.
- `1-` = Gives the probability of at least 8 successes.

### **Step 4: Final calculation**

\\[ P(X\geq 8)=1-0.5 \\]

8.  How does log transformation help in stabilizing variance and making data more normally distributed

Ans:- ## **How Does Log Transformation Help in Stabilizing Variance and Making Data More Normally Distributed?**

### **1. What is Log Transformation?**

**Log transformation** is a data transformation technique in which we replace the original values of a dataset with their logarithms.

It is commonly used when data is highly skewed or has unequal variance.

The formula is:

\\[ Y=\log(X) \\]

Where:

- \\(X\\) = Original data value
- \\(Y\\) = Log-transformed value

If the dataset contains zero values, we commonly use:

\\[ Y=\log(X+1) \\]

### **2. How Does Log Transformation Stabilize Variance?**

Variance measures how much data values are spread around the mean.

Sometimes, a dataset has larger variation when its values are larger. This is called **heteroscedasticity (unequal variance)**.

Log transformation helps by:

- Reducing the effect of very large values.
- Compressing the scale of the data.
- Reducing differences in spread across different levels.
- Making variance more consistent.

**Example:**

Consider the following values:

| Original data | Log-transformed (base 10) |
| ------------- | ------------------------- |
| 10            | 1.000                     |
| 100           | 2.000                     |
| 1,000         | 3.000                     |
| 10,000        | 4.000                     |
| 100,000       | 5.000                     |

Notice that large values become smaller on the logarithmic scale.

This reduces the influence of extremely large observations.

### **3. How Does Log Transformation Make Data More Normally Distributed?**

Many real-world datasets, such as income, population, sales, and house prices, are positively skewed (right-skewed).

Log transformation helps by:

1. Compressing large values.
2. Reducing right-skewness.
3. Bringing extreme values closer to the centre.
4. Sometimes making the distribution more symmetrical and bell-shaped.
