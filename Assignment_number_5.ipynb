1.  : What is a null hypothesis (H₀) and why is it important in hypothesis testing?

Ans:- ## **What is a Null Hypothesis (H₀)?**

A **null hypothesis (H₀)** is a statement in statistics that assumes there is **no significant effect, difference, or relationship** between the variables being studied.

It is the initial assumption made in hypothesis testing. Researchers use statistical tests to determine whether there is enough evidence to reject this assumption.

### **Example**

Suppose a researcher wants to find out whether a new teaching method improves students' exam scores.

- **Null hypothesis (H₀):** The new teaching method has no significant effect on students' exam scores.
- **Alternative hypothesis (H₁):** The new teaching method significantly improves students' exam scores.

### **Why is the Null Hypothesis Important?**

1\. **Provides a starting point:** It gives researchers an initial assumption to test.

2\. **Helps make decisions:** It helps determine whether the observed results are statistically significant.

3\. **Reduces bias:** It provides an objective basis for analyzing research results rather than relying on personal opinions.

4\. **Supports scientific research:** It helps researchers evaluate claims using statistical evidence.

5\. **Enables comparison:** It allows researchers to compare observed results with what would be expected if no effect or difference existed.

### **What happens after hypothesis testing?**

There are two possible decisions:

- **Reject H₀:** There is sufficient statistical evidence against the null hypothesis.
- **Fail to reject H₀:** There is insufficient statistical evidence against the null hypothesis.

*Note:* Failing to reject H₀ does not prove that it is true.



2.  What does the significance level (α) represent in hypothesis testing?

Ans:- ## **What is the Significance Level (α) in Hypothesis Testing?**

The **significance level (α)** is the probability of rejecting the null hypothesis (H₀) when it is actually true.

In simple words, it represents the **maximum acceptable risk of making a Type I error** (false positive) in hypothesis testing.

### **Common Significance Levels**

The commonly used significance levels are:

- **0.10 (10%)** – 10% risk of a Type I error.
- **0.05 (5%)** – 5% risk of a Type I error.
- **0.01 (1%)** – 1% risk of a Type I error.

### **Example**

Suppose a researcher chooses a significance level of **α = 0.05 (5%)** to test whether a new medicine is effective.

This means the researcher accepts a 5% risk of rejecting the null hypothesis when it is actually true.

### **How is α used in hypothesis testing?**

The significance level is compared with the p-value:

- If **p-value ≤ α:** Reject the null hypothesis (H₀).
- If **p-value > α:** Fail to reject the null hypothesis (H₀).

### **Why is the Significance Level Important?**

1. It controls the probability of a Type I error.
2. It helps researchers decide whether results are statistically significant.
3. It provides a standard for making objective decisions.
4. It helps maintain reliability in statistical conclusions.



3.  Differentiate between Type I and Type II errors.

Ans:- ## **Difference Between Type I and Type II Errors**

In hypothesis testing, **Type I and Type II errors** are two types of mistakes that can occur when making decisions about the null hypothesis (H₀).

### **1. Type I Error (α)**

A Type I error occurs when we **reject the null hypothesis (H₀) even though it is actually true.**

- It is also called a **False Positive**.
- Its probability is represented by α (significance level).
- Example: Concluding that a new medicine is effective when it actually has no effect.

### **2. Type II Error (β)**

A Type II error occurs when we **fail to reject the null hypothesis (H₀) even though it is actually false.**

- It is also called a **False Negative**.
- Its probability is represented by β.
- Example: Concluding that there is insufficient evidence that a new medicine works when it is actually effective.

### **Difference Between Type I and Type II Errors**

| Basis       | Type I Error                            | Type II Error                                                          |
| ----------- | --------------------------------------- | ---------------------------------------------------------------------- |
| Meaning     | Rejecting a true H₀                     | Failing to reject a false H₀                                           |
| Also called | False Positive                          | False Negative                                                         |
| Symbol      | α (Alpha)                               | β (Beta)                                                               |
| Probability | Significance level                      | Probability of Type II error                                           |
| Result      | Incorrectly detects an effect           | Fails to detect an actual effect                                       |
| Example     | Saying a medicine works when it doesn't | Saying there is insufficient evidence when the medicine actually works |

### **Simple Example**

Suppose a doctor tests whether a patient has a disease.

- **Type I Error:** The test says the patient has the disease, but the patient is actually healthy.
- **Type II Error:** The test says the patient does not have the disease, but the patient actually has it.



4.  Explain the difference between a one-tailed and two-tailed test. Give an example of each.

Ans:- ## **Difference Between One-Tailed and Two-Tailed Tests**

In hypothesis testing, one-tailed and two-tailed tests are used to determine whether there is a statistically significant difference or effect.

### **1. One-Tailed Test**

A **one-tailed test** is a statistical test in which the rejection region is located in only one direction (either left or right) of the probability distribution.

It is used when the researcher wants to test whether a value is specifically greater than or less than a particular value.

**Example:**

A researcher wants to test whether a new teaching method increases students' average marks.

- Null hypothesis (H₀): The new teaching method does not increase average marks.
- Alternative hypothesis (H₁): The new teaching method increases average marks.

Here, the test is one-tailed because the researcher is interested only in an increase.

A **two-tailed test** is a statistical test in which the rejection regions are located in both directions (left and right) of the probability distribution.

It is used when the researcher wants to determine whether a value is significantly different, either higher or lower, than a particular value.

A researcher wants to test whether a new teaching method changes students' average marks.

- Null hypothesis (H₀): The new teaching method does not change average marks.
- Alternative hypothesis (H₁): The new teaching method changes average marks.

Here, the test is two-tailed because the researcher is interested in both increases and decreases.

5.   A company claims that the average time to resolve a customer complaint is 10 minutes.

A random sample of 9 complaints gives an average time of 12 minutes and a standard deviation of 3 minutes.

At α = 0.05, test the claim

Ans:- ## **Hypothesis Testing: Testing the Company's Claim**

**Given data:**

- Population mean (\\(\mu\\)) = 10 minutes
- Sample mean (\\(\bar{x}\\)) = 12 minutes
- Sample standard deviation (\\(s\\)) = 3 minutes
- Sample size (\\(n\\)) = 9
- Significance level (\\(\alpha\\)) = 0.05

We need to test whether the company's claim that the average complaint resolution time is 10 minutes is correct.

### **Step 1: State the Hypotheses**

- **Null hypothesis (H₀):** \\(\mu = 10\\) minutes.
- **Alternative hypothesis (H₁):** \\(\mu \neq 10\\) minutes.

Since we are testing whether the average time is different from 10 minutes, this is a **two-tailed test**.

### **Step 2: Select the Test**

Since the sample size is small (n = 9) and the population standard deviation is unknown, we use the **one-sample t-test**.

### **Step 3: Calculate the Test Statistic**

Formula:

\\[ t=\frac{\bar{x}-\mu}{s/\sqrt{n}} \\]

Substitute the given values:

\\[ t=\frac{12-10}{3/\sqrt{9}} \\]

\\[ t=\frac{2}{3/3} \\]

\\[ \boxed{t=2.00} \\]

### **Step 4: Find the Critical Value**

- Significance level: \\(\alpha=0.05\\)
- Degrees of freedom:

\\[ df=n-1=9-1=8 \\]

For a two-tailed test at 5% significance level and 8 degrees of freedom:

\\[ \boxed{t\_{\text{critical}}=\pm 2.306} \\]

### **Step 5: Make the Decision**

Comparison

- Calculated t-value = 2.00
- Critical t-value = ±2.306

Decision rule:

- If calculated t-value is greater than 2.306 or less than -2.306, reject H₀.
- Otherwise, fail to reject H₀.

Since:

\\[ -2.306 < 2.00 < 2.306 \\]

We **fail to reject the null hypothesis (H₀).**

### **Step 6: Conclusion**

At the 5% significance level, there is insufficient statistical evidence to conclude that the average complaint resolution time is different from 10 minutes.

Therefore, the company's claim that the average resolution time is 10 minutes is **not rejected**.



6.   When should you use a Z-test instead of a t-test?

Ans:- ## **When Should You Use a Z-Test Instead of a t-Test?**

A **Z-test** is a statistical test used to determine whether there is a significant difference between a sample statistic and a population parameter, particularly when the population standard deviation is known.

A **t-test** is generally used when the population standard deviation is unknown and is estimated using the sample standard deviation.

### **Conditions for Using a Z-Test**

A Z-test is preferred in the following situations:

1. **Population standard deviation is known:** When the population standard deviation (σ) is available.
2. **Large sample size:** Generally, when the sample size is 30 or more.
3. **Normally distributed population:** When the population follows a normal distribution, especially for small samples.
4. **Testing population means or proportions:** Z-tests can be used to test population means and proportions under appropriate conditions.

### **Difference Between Z-Test and t-Test**

| Basis                         | Z-Test                        | t-Test               |
| ----------------------------- | ----------------------------- | -------------------- |
| Population standard deviation | Known                         | Unknown              |
| Sample size                   | Generally large (n ≥ 30)      | Often small (n < 30) |
| Distribution used             | Normal distribution           | t-distribution       |
| Standard deviation            | Population SD (σ)             | Sample SD (s)        |
| Application                   | Testing means and proportions | Mainly testing means |
| Degrees of freedom            | Not required                  | Required             |

### **Example of a Z-Test**

A company claims that the average weight of its product is 50 grams.

- Population mean = 50 grams
- Sample mean = 52 grams
- Population standard deviation = 5 grams
- Sample size = 100

Since the population standard deviation is known and the sample size is large, we use a **Z-test**.

### **Example of a t-Test**

A researcher wants to test the average marks of 10 students.

- Sample mean = 75
- Sample standard deviation = 5
- Sample size = 10

Since the sample size is small and the population standard deviation is unknown, we use a **t-test** (assuming the population is approximately normal).

Ans:- ## **Question: Chi-Square Test of Independence**

A company wants to test whether **product preference is independent of gender** at a 5% level of significance.

We use the **Chi-Square Test of Independence**.

### **Step 1: Given Data**

The observed frequency table is:

| Gender    | Product A | Product B | Total   |
| --------- | --------- | --------- | ------- |
| Male      | 30        | 20        | 50      |
| Female    | 10        | 40        | 50      |
| **Total** | **40**    | **60**    | **100** |

- Significance level: \\(\alpha=0.05\\)
- Total observations: \\(N=100\\)

### **Step 2: State the Hypotheses**

- **Null hypothesis (H₀):** Product preference is independent of gender.
- **Alternative hypothesis (H₁):** Product preference is dependent on gender.

### **Step 3: Calculate Expected Frequencies**

Formula:

\\[ E=\frac{\text{Row Total}\times\text{Column Total}}{\text{Grand Total}} \\]

For Male and Product A:

\\[ E=\frac{50\times40}{100}=20 \\]

For Male and Product B:

\\[ E=\frac{50\times60}{100}=30 \\]

For Female and Product A:

\\[ E=\frac{50\times40}{100}=20 \\]

For Female and Product B:

\\[ E=\frac{50\times60}{100}=30 \\]

Expected frequency table:

| Gender | Product A | Product B |
| ------ | --------- | --------- |
| Male   | 20        | 30        |
| Female | 20        | 30        |

### **Step 4: Calculate Chi-Square Value**

Formula:

\\[ \chi^2=\sum\frac{(O-E)^2}{E} \\]

Substituting the values:

| Gender | Product   | O  | E  | (O−E)²/E  |
| ------ | --------- | -- | -- | --------- |
| Male   | A         | 30 | 20 | 5         |
| Male   | B         | 20 | 30 | 3.33      |
| Female | A         | 10 | 20 | 5         |
| Female | B         | 40 | 30 | 3.33      |
|        | **Total** |    |    | **16.66** |

Therefore,

\\[ \boxed{\chi^2=16.67} \\]

### **Step 5: Calculate Degrees of Freedom**

Formula:

\\[ df=(r-1)(c-1) \\]

Where:

- r = Number of rows = 2
- c = Number of columns = 2

\\[ df=(2-1)(2-1) \\]

\\[ \boxed{df=1} \\]

At 5% significance level and 1 degree of freedom:

\\[ \boxed{\chi^2\_{\text{critical}}=3.841} \\]

### **Step 6: Make the Decision**

Compare calculated and critical values:

- Calculated value = 16.67
- Critical value = 3.841

Decision rule:

- If calculated value is greater than critical value, reject H₀.
- If calculated value is less than critical value, fail to reject H₀.

Since:

\\[ 16.67 > 3.841 \\]

Reject H₀

### **Step 7: Conclusion**

At the 5% level of significance, we reject the null hypothesis.

Therefore, **product preference is not independent of gender.** There is a statistically significant association between gender and product preference.
