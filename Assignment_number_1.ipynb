1.Explain the difference between Absolute, Relative, and Mixed Cell
Referencing in Excel with examples.

Ans:- ## Cell Referencing in Excel

Cell referencing in Microsoft Excel means identifying a cell or a range of cells in a formula. Excel mainly uses three types of cell referencing: **Relative, Absolute, and Mixed Cell Referencing**.

### 1. Relative Cell Referencing

In **relative cell referencing**, the cell reference changes automatically when the formula is copied or moved to another cell.

**Example:**

Suppose cell **A1** contains 10 and cell **B1** contains 20. In cell C1, we enter:

`=A1+B1`

If we copy this formula from C1 to C2, Excel automatically changes it to:

`=A2+B2`

Thus, relative references adjust according to the position of the formula.

**Use:** It is useful when the same calculation needs to be performed for different rows or columns.

---

### 2. Absolute Cell Referencing

In **absolute cell referencing**, the cell reference remains fixed even when the formula is copied to another cell. A dollar sign (`$`) is used before both the column and row.

**Example:**

Suppose cell **A1** contains a price of ₹1,000 and cell **B1** contains a tax rate of 10%. We can use:

`=A1*$B$1`

When this formula is copied down, `$B$1` remains fixed. Only A1 changes to A2, A3, and so on.

**Use:** It is useful when a formula needs to refer to a constant value, such as a tax rate, discount rate, or fixed percentage.

---

### 3. Mixed Cell Referencing

In **mixed cell referencing**, either the row or the column is fixed, while the other part changes when the formula is copied. There are two types:

* **$A1** – Column A is fixed, but the row changes.
* **A$1** – Row 1 is fixed, but the column changes.

**Example:**

If cell A1 contains a value and the formula is:

`=$A1*B$1`

When the formula is copied to another cell, **column A** remains fixed in `$A1`, while the row can change. Similarly, **row 1** remains fixed in `B$1`, while the column can change.

**Use:** Mixed references are useful for calculations involving tables where either a particular row or column must remain constant.

### Difference at a Glance

| Type         | Example        | What Changes When Copied?    |
| ------------ | -------------- | ---------------------------- |
| **Relative** | `A1`           | Row and column both change   |
| **Absolute** | `$A$1`         | Nothing changes              |
| **Mixed**    | `$A1` or `A$1` | Either row or column changes |

3 : What are Text Functions in Excel? Mention any five with examples.

Ans:-
**Text Functions in Excel** are built-in functions used to manipulate, format, combine, and extract text from cells. They are useful for cleaning and organizing textual data.

### Five Text Functions with Examples

1. **LEFT()**
   The LEFT function extracts a specified number of characters from the beginning of a text string.
   **Example:** `=LEFT("COMPUTER",4)`
   **Result:** `COMP`

2. **RIGHT()**
   The RIGHT function extracts a specified number of characters from the end of a text string.
   **Example:** `=RIGHT("COMPUTER",4)`
   **Result:** `UTER`

3. **MID()**
   The MID function extracts characters from the middle of a text string, starting from a specified position.
   **Example:** `=MID("COMPUTER",3,4)`
   **Result:** `MPUT`

4. **LEN()**
   The LEN function counts the total number of characters in a text string, including spaces.
   **Example:** `=LEN("EXCEL")`
   **Result:** `5`

5. **UPPER()**
   The UPPER function converts text into uppercase letters.
   **Example:** `=UPPER("excel functions")`
   **Result:** `EXCEL FUNCTION'

It is useful in decision making because it allows users to create and compare multiple scenarios, such as **Best Case, Worst Case, and Most Likely Case**, without changing the original data permanently.

4. Helps identify the risks and opportunities associated with different situations.

from XLOOKUP? Which among XLOOKUP and INDEX-MATCH is best while usage

Ans:- ## VLOOKUP, HLOOKUP, XLOOKUP, and INDEX-MATCH

### 1. VLOOKUP

**VLOOKUP (Vertical Lookup)** is used to search for a value in the **first column of a table** and return a corresponding value from another column in the same row.

**Example:**

`=VLOOKUP(101,A2:D10,3,FALSE)`

This searches for **101** in the first column of the range A2:D10 and returns the value from the **third column**.

**Purpose:** It is mainly used for searching data arranged vertically.

---

### 2. HLOOKUP

**HLOOKUP (Horizontal Lookup)** searches for a value in the **first row of a table** and returns a corresponding value from another row in the same column.

**Example:**

`=HLOOKUP("March",A1:F4,3,FALSE)`

This searches for **March** in the first row and returns the corresponding value from the **third row**.

**Purpose:** It is mainly used for searching data arranged horizontally.

---

### 3. XLOOKUP

**XLOOKUP** is a newer and more flexible lookup function in Excel. It can search both **vertically and horizontally** and can return values from any direction.

**Example:**

`=XLOOKUP(101,A2:A10,C2:C10,"Not Found")`

This searches for **101** in A2:A10 and returns the corresponding value from C2:C10.

### Difference Between VLOOKUP/HLOOKUP and XLOOKUP

| Feature               | VLOOKUP     | HLOOKUP           | XLOOKUP                     |
| --------------------- | ----------- | ----------------- | --------------------------- |
| Search direction      | Vertical    | Horizontal        | Both                        |
| Can look to the left? | No          | Limited by layout | Yes                         |
| Default match         | Approximate | Approximate       | Exact                       |
| If value is not found | Error       | Error             | Can specify a custom result |
| Flexibility           | Limited     | Limited           | High                        |

---

### 4. XLOOKUP vs. INDEX-MATCH

**INDEX-MATCH** is a combination of two Excel functions. `MATCH` finds the position of a value, while `INDEX` returns the value from that position.

**Example:**

`=INDEX(C2:C10,MATCH(101,A2:A10,0))`

XLOOKUP can perform the same type of lookup more simply:

`=XLOOKUP(101,A2:A10,C2:C10)`

### Which is Better: XLOOKUP or INDEX-MATCH?

For **most new Excel work, XLOOKUP is the better choice** because it is easier to write, easier to understand, supports exact matching naturally, can look in either direction, and allows a custom "not found" result.

However, **INDEX-MATCH is still useful** when working with older Excel versions that do not support XLOOKUP or in existing workbooks that already use INDEX-MATCH.

Ans:- ## Product Stock Management Using Macro in Excel

### 1. Create the Product Table

Create the following table in Excel:

| Product ID | Product Name | Stock Level |
| ---------- | ------------ | ----------: |
| P001       | Laptop       |           8 |
| P002       | Keyboard     |          25 |
| P003       | Mouse        |          55 |
| P004       | Monitor      |          12 |
| P005       | Printer      |          65 |
| P006       | Headphones   |           7 |
| P007       | Webcam       |          35 |
| P008       | USB Drive    |           5 |
| P009       | Speaker      |          75 |
| P010       | Router       |          18 |

### 2. Create the Macro

The Macro will check the stock level and automatically highlight:

* **Less than 10 units → Red**
* **More than 50 units → Green**
* **10 to 50 units → No special highlighting**

Open Excel and press **Alt + F11** to open the VBA Editor.

Then select:

**Insert → Module**

Enter the following VBA code:

```vba
Sub HighlightStock()

    Dim ws As Worksheet
    Dim lastRow As Long
    Dim i As Long
    Dim stock As Long

    Set ws = ActiveSheet
    lastRow = ws.Cells(ws.Rows.Count, "C").End(xlUp).Row

    For i = 2 To lastRow

        stock = ws.Cells(i, 3).Value

        If stock < 10 Then
            ws.Cells(i, 3).Interior.Color = RGB(255, 0, 0)
        
        ElseIf stock > 50 Then
            ws.Cells(i, 3).Interior.Color = RGB(0, 255, 0)
        
        Else
            ws.Cells(i, 3).Interior.ColorIndex = xlNone
        End If

    Next i

End Sub
```

### 3. Run the Macro

1. Close the VBA Editor.
2. Return to Excel.
3. Press **Alt + F8**.
4. Select **HighlightStock**.
5. Click **Run**.

The Stock Level column will now be highlighted automatically.

Ans:- ## Drop-Down List of Products and Automatic Price Display

The given dataset contains product details, including **Product Name** and **Unit Price (₹)**. The objective is to create a drop-down list of product names and automatically display the corresponding price when a product is selected.

### Steps:

1. Select an empty cell, for example **K2**, and enter the heading **Select Product**.

2. Select cell **K3** and go to:
   **Data → Data Validation → Add Rule**.

3. Under **Criteria**, select **Dropdown (from a range)**.

4. Select the **Product Name** range from the dataset:
   `G2:G50`

5. Click **Done**. A drop-down list of product names will now appear in K3.

6. In the adjacent cell **L2**, enter the heading **Price**.

7. In **L3**, enter the following XLOOKUP formula:

```excel
=XLOOKUP(K3,G:G,I:I,"Product Not Found")
```

Here:

* `K3` = selected product from the drop-down
* `G:G` = Product Name column
* `I:I` = Unit Price column
* `"Product Not Found"` = message displayed if the product does not exist

### Example

If **Smartwatch** is selected from the drop-down, the adjacent cell will automatically display its corresponding price, such as **₹999**.

| Select Product | Price |
| -------------- | ----: |
| Smartwatch ▼   |  ₹999 |

If another product is selected, its corresponding price will automatically change.

### Alternative Formula

If XLOOKUP is not available, the following VLOOKUP formula can be used:

```excel
=VLOOKUP(K3,G:I,3,FALSE)
```
