1.  A die is rolled. What is the probability of getting:

(a) An even number

(b) A number greater than 4

Ans:- A die has **6 possible outcomes**: 1, 2, 3, 4, 5, 6.

Formula:

\\[ P(E)=\frac{\text{Favourable outcomes}}{\text{Total outcomes}} \\]

**(a) Probability of getting an even number**

Even numbers are 2, 4, 6.

Number of favourable outcomes = 3

\\[ P(\text{Even number})=\frac{3}{6} =\boxed{\frac{1}{2}} \\]

**(b) Probability of getting a number greater than 4**

Numbers greater than 4 are 5, 6.

Number of favourable outcomes = 2

\\[ P(\text{Number greater than 4})=\frac{2}{6} =\boxed{\frac{1}{3}} \\]

**Final Answers:**

- (a) \\(\boxed{\frac{1}{2}}\\)
- (b) \\(\boxed{\frac{1}{3}}\\)

2.  In a class of 50 students:

20 like Mathematics (M)

15 like Science (S)

5 like both subjects

What is the probability that a student chosen at random likes Mathematics or Science?

Ans:- Given:

- Total students = 50
- Students who like Mathematics (M) = 20
- Students who like Science (S) = 15
- Students who like both subjects = 5

**Formula:**

\\[ P(M\cup S)=\frac{n(M)+n(S)-n(M\cap S)}{n(U)} \\]

**Step 1: Substitute the values.**

\\[ P(M\cup S)=\frac{20+15-5}{50} \\]

**Step 2: Simplify.**

\\[ =\frac{30}{50} \\]

\\[ =\boxed{\frac{3}{5}} \\]

Final Answer

\\[ \boxed{P(M\cup S)=\frac{3}{5}=0.6=60\\%} \\]

Therefore, the probability that a randomly chosen student likes Mathematics or Science is **60%**.

3. A bag has 3 red and 2 blue balls. If one ball is drawn randomly and is red, what is the probability
that the next ball is also red (without replacement)?


Ans:- Given:

- Red balls = 3
- Blue balls = 2
- Total balls = 5

The first ball drawn is **red**.

Since there is no replacement, one red ball is removed.

**Step 1: Find the remaining balls.**

- Red balls = 2
- Blue balls = 2
- Total remaining balls = 4

**Step 2: Calculate the probability of getting another red ball.**

\\[ P(\text{Red on second draw})= \frac{\text{Remaining red balls}}{\text{Total remaining balls}} \\]

\\[ =\frac{2}{4} \\]

\\[ =\boxed{\frac{1}{2}} \\]

Final Answer

\\[ \boxed{\frac{1}{2}=50\\%} \\]

Therefore, the probability that the next ball is also red is **1/2 or 50%**.

4. : The population of a school is divided into 60% boys and 40% girls. If you want equal representation
of both genders in the sample, which method should you use: Simple Random Sampling or Stratified Sampling?
Why?

Ans:- **Answer: Stratified Sampling.**

**Why?**

The school population consists of:

- 60% Boys
- 40% Girls

If we use **Simple Random Sampling**, we may not get an equal number of boys and girls in the sample.

In **Stratified Sampling**, the population is divided into two groups (strata):

1. Boys
2. Girls

Then, an equal number of students is selected from each group to ensure equal representation.

**Conclusion:** Stratified Sampling is the better method because it ensures that both boys and girls are equally represented in the sample.

5.The average height of 1000 students = 160 cm. A sample of 100 students shows an average height
= 158 cm.

Find the sampling error

Ans:- **Given:**

- Population mean = 160 cm
- Sample mean = 158 cm

**Formula:**

Sampling Error = Sample Mean − Population Mean

**Step 1: Substitute the values.**

\\[ =158-160 \\]

\\[ =-2\text{ cm} \\]

**Final Answer:**

\\[ \boxed{\text{Sampling Error}=-2\text{ cm}} \\]

Therefore, the sampling error is **2 cm** (the sample average is 2 cm lower than the population average).

6.  The population mean salary is ₹50,000 with σ = ₹5,000. If we take a sample of 100 employees, what
is the standard error of the mean (SEM)?


Ans:- **Given:**

- Population standard deviation (\\(\sigma\\)) = ₹5,000
- Sample size (\\(n\\)) = 100

**Formula:**

Standard Error of Mean (SEM):

\\[ SEM=\frac{\sigma}{\sqrt{n}} \\]

**Step 1: Substitute the values.**

\\[ SEM=\frac{5000}{\sqrt{100}} \\]

**Step 2: Calculate.**

\\[ =\frac{5000}{10} \\]

\\[ =\boxed{₹500} \\]

Final Answer

**Standard Error of the Mean (SEM) = ₹500.**

8. : From a deck of 52 cards, two cards are drawn without replacement. What is the probability that
both are Aces?

Ans:- **Given:**

- Total cards = 52
- Number of Aces = 4
- Two cards are drawn without replacement.

**Formula:**

\\[ P(\text{Both Aces})=P(A_1)\times P(A_2) \\]

**Step 1: Probability of drawing the first Ace.**

\\[ P(A_1)=\frac{4}{52} \\]

**Step 2: Probability of drawing the second Ace.**

Since the first Ace is not replaced:

- Remaining Aces = 3
- Remaining cards = 51

\\[ P(A_2)=\frac{3}{51} \\]

**Step 3: Multiply the probabilities.**

\\[ P(\text{Both Aces})=\frac{4}{52}\times\frac{3}{51} \\]

\\[ =\frac{12}{2652} \\]

\\[ =\boxed{\frac{1}{221}} \\]

Final Answer

\\[ \boxed{\frac{1}{221}\approx 0.00452} \\]

Therefore, the probability of drawing two Aces is **1/221 (approximately 0.452%)**.

10.   Differentiate between discrete and continuous random variables with examples.

Ans:- ## **Difference Between Discrete and Continuous Random Variables**

A **random variable** is a variable whose value depends on the outcome of a random experiment.

| Discrete Random Variable                | Continuous Random Variable                          |
| --------------------------------------- | --------------------------------------------------- |
| Takes countable values.                 | Takes any value within a given range.               |
| Usually obtained by counting.           | Usually obtained by measuring.                      |
| Values may be whole numbers.            | Values may include decimals and fractions.          |
| Has separate, distinct values.          | Can take infinitely many values within an interval. |
| Example: Number of students in a class. | Example: Height or weight of a student.             |

### **1. Discrete Random Variable**

A discrete random variable takes **countable and distinct values**.

**Examples:**

- Number of heads when tossing a coin.
- Number of students in a classroom.
- Number of defective bulbs in a factory.
- Number of cars in a parking area.

### **2. Continuous Random Variable**

A continuous random variable can take **any value within a specific range**.

**Examples:**

- Height of a person (160.5 cm).
- Weight of a student (55.7 kg).
- Temperature (30.5°C).
- Time taken to complete an exam (2.5 hours).

### **Main Difference to Remember**

**Discrete = Counting**

# **1, 2, 3, 4, 5...**

Example: Number of students

**Continuous = Measuring**

# **1.2, 1.5, 1.75, 2.3...**

Example: Height, weight and time

**Conclusion:** Discrete random variables are based on counting, whereas continuous random variables are based on measurement.
